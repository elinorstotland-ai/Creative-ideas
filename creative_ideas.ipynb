# Creative Ideas: A New Use for an Everyday Object

Enter an object you have around the house. This notebook asks OpenAI for **one unusual, practical way to use it**.

Your API key is requested privately when you run the setup cell. It is not saved in this notebook.

## 2. Enter your API key

Get an API key from your OpenAI account, then enter it when prompted. getpass hides what you type. Do not paste the key into a code cell.

```python
from getpass import getpass
from openai import OpenAI

# Type your API key into the private prompt. It will not be displayed.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
```

## 3. Choose an everyday object

Try something simple, such as a rubber band, mug, or cardboard box.

```python
object_name = input("Name an everyday object: ").strip()

if not object_name:
    raise ValueError("Please enter an object name and run this cell again.")

print(f"Your object: {object_name}")
```

## 4. Get one creative idea

The request asks for exactly one idea that is both unusual and practical.

```python
response = client.responses.create(
    model="gpt-6-luna",
    input=(
        f"Suggest exactly ONE unusual but practical use for a {object_name}. "
        "Explain briefly how to use it. Keep the suggestion safe and beginner-friendly."
    ),
)

print(response.output_text)
```
