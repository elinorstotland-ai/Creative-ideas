# Creative Ideas: Three Uses for an Everyday Object

Enter an object you have around the house. This notebook asks OpenAI for **three unusual, practical ways to use it**. You can keep trying new objects until you type **exit**.

Your API key is requested privately when you run the setup cell. It is not saved in this notebook.

## 2. Enter your API key

Get an API key from your OpenAI account, then enter it when prompted. getpass hides what you type. Do not paste the key into a code cell.

```python
from getpass import getpass
from openai import OpenAI

# Type your API key into the private prompt. It will not be displayed.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
```

## 3. Enter objects to explore

Type an object such as a rubber band, mug, or cardboard box. After the ideas appear, enter another object to continue, or type exit to stop.

```python
# Keep asking for objects until you choose to stop.
while True:
    object_name = input("Name an everyday object (or type 'exit' to stop): ").strip()

    # Stop the loop when the user types exit.
    if object_name.lower() == "exit":
        print("Thanks for exploring creative ideas!")
        break

    # Ask again if the input is blank.
    if not object_name:
        print("Please enter an object name, or type 'exit' to stop.")
        continue

    # Make one API request for exactly three distinct ideas.
    response = client.responses.create(
        model="gpt-6-luna",
        input=(
            f"Suggest exactly THREE distinct, unusual, practical uses for a {object_name}. "
            "Format them as a numbered list with exactly three items labeled 1, 2, and 3. "
            "Briefly explain each idea and keep all suggestions safe and beginner-friendly."
        ),
    )

    # Show the model's response, then the loop asks for another object.
    print(f"\nThree ideas for {object_name}:\n")
    print(response.output_text)
    print()
```
